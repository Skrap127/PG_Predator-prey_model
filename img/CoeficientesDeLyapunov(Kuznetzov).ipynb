# Cálculo del primer y segundo coeficiente de Lyapunov (Kuznetzov)
## Primer coeficiente de Lyapunov

```
PP[u_, v_, A_, Q_] := ((1-u)(u^2+A^2)-Q*u*v)u^2;
QQ[u_, v_, A_, B_] := B(u-v)(u^2+A^2)v;
```

```
asumir = {a > 0, b < 0, d > 0, m > 0, 0 < A < (H* Sqrt[1 - 2*H]), 0 < H < (1/2), B > 0, Q > 0, omega > 0};

h1[u_, v_, A_, Q_] = PP[u + H, v + H, A, Q] ;
h2[u_, v_, A_, B_] = QQ[u + H, v + H, A, B] ;
F1[u_, v_, A_, Q_] = h1[u, v, A, Q] - a*u - b*v ; 
F2[u_, v_, A_, B_] = h2[u, v, A, B] - a*u + a*v ;

p = {I/(2*m), (1/2) - I/(2*m)};
q = {1+m*I, 1};
```

```
operarComplejos[expr_] := FullSimplify[Assuming[asumir, Refine[expr]]];
primerCom = operarComplejos[z*q[[1]] + w*Conjugate[q[[1]]]];
segundoCom = operarComplejos[z*q[[2]] + w*Conjugate[q[[2]]]];
```

```
GG = Series[operarComplejos[Conjugate[p[[1]]]*F1[primerCom, segundoCom, A, Q] + Conjugate[p[[2]]]*F2[primerCom, segundoCom, A, B]], {z, 0, 6}, {w, 0, 6}];
G[z_, w_] = Normal[GG]
```

El primer coeficiente de Lyapunov está dada por la expresión: $$l_1​(0)=\frac{1}{2 \omega_0 ^2}​ Re(i g_{20}​g_{11}​+\omega_0​ g_{21}​),$$ donde:

```
g20 = FullSimplify[Derivative[2,0][G][0,0], asumir]
g11 = FullSimplify[Derivative[1,1][G][0,0], asumir]
g21 = FullSimplify[Derivative[2,1][G][0,0], asumir]
```

```
product1 = FullSimplify[I*g20*g11, asumir];
product2 = omega*g21;
parentesis = FullSimplify[Re[ComplexExpand[product1+product2]], asumir];
```

```
ll1 = FullSimplify[(1/(2*omega^2))*parentesis, asumir];
```

```
B = (H^2(1 - 2H) - A^2)/(A^2 + H^2);
Q = (1 - H)(A^2 + H^2)/H^2;
a = H(H^2(1-2H)-A^2);
b = -H((1-H)(A^2+H^2));
d = H(H^3 + A^2*(2 - H));
m = Sqrt[d/a];
omega = Sqrt[a*d];

l1 = FullSimplify[ll1, asumir]
```

```
Clear[A, B, Q, a, b, d, m, n, omega]
```

## Segundo coeficiente de Lyapunov
El segundo coeficiente de Lyapunov está dada por la expresión:
$$
\begin{aligned}
12l_2(0) &= \frac{1}{\omega_0}\text{Re } g_{32} \\
&+ \frac{1}{\omega_0^2}\text{Im } \left[g_{20}\bar{g}_{31} - g_{11}(4g_{31} + 3\bar{g}_{22}) - \frac{1}{3}g_{02}(g_{40} + \bar{g}_{13}) - g_{30}g_{12}\right] \\
&+ \frac{1}{\omega_0^3} \left\{\text{Re } \left[g_{20}(\bar{g}_{11}(3g_{12} - \bar{g}_{30}) + g_{02}\left(\bar{g}_{12} - \frac{1}{3}g_{30}\right) + \frac{1}{3}\bar{g}_{02}g_{03})\right.\right. \\
&\quad\quad\quad + g_{11}(\bar{g}_{02}\left(\frac{5}{3}\bar{g}_{30} + 3g_{12}\right) + \frac{1}{3}g_{02}\bar{g}_{03} - 4g_{11}g_{30})\Big] \\
&\quad\quad\quad + 3\left.\text{Im}(g_{20}g_{11})\text{ Im } g_{21} \right\} \\
&+ \frac{1}{\omega_0^4} \left\{\text{Im}\left[g_{11}\bar{g}_{02}\left(\bar{g}_{20}^2 - 3\bar{g}_{20}g_{11} - 4g_{11}^2\right)\right]\right. \\
&\quad\quad\quad + \left.\text{Im}(g_{20}g_{11})\left[3\text{ Re}(g_{20}g_{11}) - 2|g_{02}|^2\right]\right\},
\end{aligned}
$$
donde:

```
g20 = FullSimplify[Derivative[2, 0][G][0, 0], asumir];
g11 = FullSimplify[Derivative[1, 1][G][0, 0], asumir];
g31 = FullSimplify[Derivative[3, 1][G][0, 0], asumir]
g32 = FullSimplify[Derivative[3, 2][G][0, 0], asumir]
g22 = FullSimplify[Derivative[2, 2][G][0, 0], asumir]
g02 = FullSimplify[Derivative[0, 2][G][0, 0], asumir]
g40 = FullSimplify[Derivative[4, 0][G][0, 0], asumir]
g13 = FullSimplify[Derivative[1, 3][G][0, 0], asumir]
g30 = FullSimplify[Derivative[3, 0][G][0, 0], asumir]
g12 = FullSimplify[Derivative[1, 2][G][0, 0], asumir]
g03 = FullSimplify[Derivative[0, 3][G][0, 0], asumir]
g21 = FullSimplify[Derivative[2, 1][G][0, 0], asumir];
```

```
asumir = {0 < H < (3 - Sqrt[6])/3, 0 < A < (H* Sqrt[1 - 2*H]), B > 0, Q >0, a > 0, b < 0, d > 0, m > 0, omega > 0};
operarComplejos[expr_] := FullSimplify[Assuming[asumir, Refine[expr]]];
```

```
cg20 = operarComplejos[Conjugate[ComplexExpand[g20]]];
cg31 = operarComplejos[Conjugate[ComplexExpand[g31]]];
cg11 = operarComplejos[Conjugate[ComplexExpand[g11]]];
cg22 = operarComplejos[Conjugate[ComplexExpand[g22]]];
cg02 = operarComplejos[Conjugate[ComplexExpand[g02]]];
cg13 = operarComplejos[Conjugate[ComplexExpand[g13]]];
cg30 = operarComplejos[Conjugate[ComplexExpand[g30]]];
cg12 = operarComplejos[Conjugate[ComplexExpand[g12]]];
cg03 = operarComplejos[Conjugate[ComplexExpand[g03]]];
```

```
factor1 = operarComplejos[Re[ComplexExpand[g32]]]
```

```
fac21 = operarComplejos[g20*cg31];
fac22 = operarComplejos[g11*(4*g31+3*cg22)];
fac23 = operarComplejos[(g02(g40+cg13))/3];
fac24 = operarComplejos[g30*g12];

factor2 = operarComplejos[Im[ComplexExpand[fac21-fac22-fac23-fac24]]]
```

```
fac31 = operarComplejos[g20*cg11*(3*g12-cg30)];
fac32 = operarComplejos[g20*g02(cg12-(g30/3))];
fac33 = operarComplejos[(g20*cg02*g03)/3];
fac34 = operarComplejos[g11*cg02(((5*cg30)/3)+3*g12)];
fac35 = operarComplejos[(g11*g02*cg03)/3];
fac36 = operarComplejos[4*g30*(g11)^2];
fac37 = operarComplejos[g20*g11];

factor3p1 = operarComplejos[Re[ComplexExpand[fac31+fac32+fac33+fac34+fac35-fac36]]]
factor3p2 = operarComplejos[Im[ComplexExpand[fac37]]]
factor3p3 = operarComplejos[Im[ComplexExpand[g21]]]
```

```
fac41 = operarComplejos[g11*cg02((cg20)^2-3*cg20*g11-4(g11)^2)];
fac42 = operarComplejos[g20*g11];
fac43 = operarComplejos[g02*cg02];

factor4p1 = operarComplejos[Im[ComplexExpand[fac41]]]
factor4p2 = operarComplejos[Im[ComplexExpand[fac42]]]
factor4p3 = operarComplejos[Re[ComplexExpand[fac42]]]
```

```
ll2 = operarComplejos[(1/omega)*factor1 + (1/omega^2)*factor2 + (1/omega^3)*(factor3p1+3*factor3p2*factor3p3) + (1/omega^4)*(factor4p1+factor4p2(3*factor4p3-2*fac43))]/12
```

```
A = Sqrt[(2*H^3 + H^2*(1 - H)*Sqrt[3*H*(4 - H)]) / (4 - 3*H*(3 - H))];
B = (2 - 3*H - Sqrt[3*H*(4 - H)])/2;
Q = ((1 - H)^2 * (4 - 3*H + Sqrt[3*H*(4 - H)])) / (4 - 3*H*(3 - H));
a = (H^3 * (1 - H) * (4 - 3*H*(5 - 2*H) - Sqrt[3*H*(4 - H)])) / (4 - 3*H*(3 - H));
b = -((H^3 * (1 - H)^2 * (4 - 3*H + Sqrt[3*H*(4 - H)])) / (4 - 3*H*(3 - H)));
d = H^4 + (H^3 * (2 - H) * (2*H + (1 - H)*Sqrt[3*H*(4 - H)])) / (4 - 3*H*(3 - H));
m = Sqrt[(H*(7 - 3*H) + (1 - H)*Sqrt[3*H*(4 - H)]) / (2 - 6*H*(2 - H))];
omega = Sqrt[a*d];

l2 = Simplify[ll2, 0 < H < (3 - Sqrt[6])/3]
```

```
lll2[h_] := l2 /. {H -> h};

example = Table[
  Module[{h, maxh},
    maxh = (3 - Sqrt[6])/3;
    h = RandomReal[{0.000001, maxh-0.000001}];
    {h, lll2[h]}
  ],
  {10}
]
```

```
plot = LogPlot[-l2, {H, 0, (3 - Sqrt[6])/3}, PlotLabel -> "Curva de Hopf generalizada", PlotTheme -> "Scientific", PlotStyle -> Black]
```
