```
r = 0.1; a = 0.1; s = 0.1; K = 0.1; n = 0.1; q = 0.1;
A = a/K;
Q = (q*n)/r;
(* B = s/r; *)
solution = NSolve[u^3 + (Q-1)u^2 + (A^2)u - (A^2) == 0, u, Reals];
H = u /. solution[[1]]
```

```
B = 0.5;

plot1 = StreamPlot[{u^2((1-u)(u^2+A^2)-Q*u*v), B(u-v)(u^2+A^2)v}, {u, -.1, 1.5}, {v, -.1, 1.5},
  Epilog -> {PointSize[Large], Point[{{0, 0}, {1, 0}, {H, H}}]},
  StreamPoints -> {{{{0.0845, 0}, Red}, Automatic}},
  StreamColorFunction -> None,
  PlotTheme -> "Scientific",
  StreamStyle -> Gray,
  StreamMarkers -> "PinDart"
]
```
