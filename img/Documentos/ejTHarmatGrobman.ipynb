```
plot = StreamPlot[{-x, y+x^2}, {x, -1, 1}, {y, −1, 1},
  Frame -> False,
  Epilog -> {PointSize[Large], Point[{{0, 0}}]},
  StreamColorFunction -> None,
  PlotTheme -> "Scientific",
  StreamStyle -> Gray,
  StreamMarkers -> "PinDart"
]
```

```
plot2 = StreamPlot[{-x, y}, {x, -1, 1}, {y, −1, 1},
  Frame -> False,
  Epilog -> {PointSize[Large], Point[{{0, 0}}]},
  StreamColorFunction -> None,
  PlotTheme -> "Scientific",
  StreamStyle -> Gray,
  StreamMarkers -> "PinDart"
]
```

```
e = Graphics[{
   Arrowheads[0.2],
   (* Flecha superior: curva hacia la derecha con H *)
   Arrow[BSplineCurve[{{0, 0.3}, {0.5, 0.5}, {1, 0.3}}]],
   Text[Style["H", 14], {0.5, 0.55}],
   
   (* Flecha inferior: curva hacia la izquierda con H^-1 *)
   Arrow[BSplineCurve[{{1, -0.3}, {0.5, -0.5}, {0, -0.3}}]],
   Text[Style[Superscript["H", -1], 14], {0.5, -0.55}]
   },
   ImageSize -> 80, Axes -> False
]
```

```
eqPo = GraphicsRow[{plot, e, plot2}]
```
