```
δ = 2;
c = 5;

h[x_] := Piecewise[{{δ*x, 0 <= x < c},{δ*c, c <= x}}]

typeI = Plot[h[x], {x, 0, 10}, PlotStyle -> {Black}, Ticks -> None,
 Epilog -> {
   {Black, PointSize[Large], Point[{0, 0}]}
 },
 PlotLabel -> "Respuesta funcional tipo 1"]
```

```
q = 2;
a = 3;

h[x_] := (q x)/(x + a)

typeII = Plot[{h[x], q}, {x, 0, 20}, PlotStyle -> {Black, {Dashed, Gray}}, 
 Axes -> True, Ticks -> None, 
 Epilog -> {
   {Black, PointSize[Large], Point[{0, 0}]}
 },
 PlotLabel -> "Respuesta funcional tipo 2"]
```

```
q = 1;
a = 4;

h[x_] := (q x^2)/(x^2 + a^2)

typeIII =Plot[{h[x],q}, {x, 0, 15}, PlotStyle -> {Black, {Dashed, Gray}}, 
 Axes -> True, Ticks -> None, 
 Epilog -> {
   {Black, PointSize[Large], Point[{0, 0}]}
 },
 PlotLabel -> "Respuesta funcional tipo 3"]
```

```
q = 1;
a = 3;

h[x_] := (q x)/(x^2 + a^2)

typeIV = Plot[h[x], {x, 0, 16}, PlotStyle -> {Black}, 
 Axes -> True, Ticks -> None, 
 Epilog -> {
   {Black, PointSize[Large], Point[{0, 0}]}
 },
 PlotLabel -> "Respuesta funcional tipo 4"]
```
